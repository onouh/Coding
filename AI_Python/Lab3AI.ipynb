```python
from dataclasses import dataclass, field
from typing import Any

@dataclass
class Node:
    score: int = 0
    move: Any = None
    children: list["Node"] = field(default_factory=list)

    def add_child(self, child: "Node") -> "Node":
        self.children.append(child)
        return child

    def add_leaf(self, score: int, move: Any = None) -> "Node":
        leaf = Node(score=score, move=move)
        self.children.append(leaf)
        return leaf

    def get_children(self, maximizing_player: bool) -> list["Node"]:
        # maximizing_player is accepted to match minimax() signature
        return self.children

    def is_terminal(self) -> bool:
        return len(self.children) == 0
```

```python
def evaluate(state):
    return state.score
```

```python
def minimax(state, depth, maximizing_player):
    if depth == 0 or is_terminal(state):
        return evaluate(state), None

    best_child = None

    if maximizing_player:
        best_score = float("-inf")
        for child in get_children(state, maximizing_player):
            score, _ = minimax(child, depth - 1, False)
            if score > best_score:
                best_score = score
                best_child = child
    else:
        best_score = float("inf")
        for child in get_children(state, maximizing_player):
            score, _ = minimax(child, depth - 1, True)
            if score < best_score:
                best_score = score
                best_child = child

    return best_score, best_child
```

```python
goals = {
    "eligible_to_graduae": ["completed_all_courses", "no_fees_due"],
    "graduate": "eligible_to_graduae" 
}

facts = {    "completed_all_courses": True,
    "no_fees_due": True  }
def prove_goal(goal):
    if goal in facts:
        return True
    elif goal in goals:
        return all(prove_goal(subgoal) for subgoal in goals[goal])
    else:
        return False
    
if prove_goal("graduate"):
    print("Congratulations! You have graduated.")
else:    print("You are not eligible to graduate yet.")
```
