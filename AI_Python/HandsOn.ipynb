```python
from experta import *

class ComputerIssueExpert(KnowledgeEngine):
    # Rule 1: Power Supply Problem 
    @Rule(Fact(no_power=True))
    def power_supply_issue_ForwardChaining(self):
        print("Forward Chaining: Diagnosis - Power Supply Problem")

    # Rule 2: Overheating or Hardware Fault 
    @Rule(Fact(blue_screen=True) & Fact(slow_performance=True))
    def hardware_fault_ForwardChaining(self):
        print("Forward Chaining: Diagnosis - Overheating or Hardware Fault")

    # Rule 3: Network Problem 
    @Rule(Fact(no_power=True))
    def power_supply_issue_ForwardChaining(self):
        print("Forward Chaining: Diagnosis - Power Supply Problem") 

    # Rule 4: Driver Issues
    @Rule(Fact(blue_screen=True))
    def driver_suggestion_ForwardChaining(self):
        print("Forward Chaining: Suggestion - Check for Driver Issues") 

    # Rule 1: Network Problem (Configured for Backward Chaining goal)
    @Rule(Fact(goal='network') & Fact(no_internet=True))
    def network_verified_BackwardChaining(self):
        print("Backward Chaining: Network Problem Verified")

    # Rule 2: Power Supply Problem (Configured for Backward Chaining goal)
    @Rule(Fact(goal='power') & Fact(no_power=True))
    def power_supply_verified_BackwardChaining(self):
        print("Backward Chaining: Power Supply Problem Verified")

    # Rule 3: Overheating or Hardware Fault (Configured for Backward Chaining goal)
    @Rule(Fact(goal='hardware') & Fact(blue_screen=True) & Fact(slow_performance=True))
    def hardware_fault_verified_BackwardChaining(self):
        print("Backward Chaining: Overheating or Hardware Fault Verified")

    # Rule 4: Driver Issues (Configured for Backward Chaining goal)
    @Rule(Fact(goal='driver') & Fact(blue_screen=True))
    def driver_issue_verified_BackwardChaining(self):
        print("Backward Chaining: Driver Issues Verified")
```

```python
# Initialize the engine
engine = ComputerIssueExpert()
engine.reset()

# Declare facts: Symptoms are present
engine.declare(
    Fact(no_power=True),
    Fact(blue_screen=True),
    Fact(slow_performance=True)
)

# Run the engine
print("--- Test Case 1 Results ---")
engine.run()
```

```python
# Reset for a new scenario
engine.reset() 

# Declare the goal and the supporting fact 
engine.declare(
    Fact(goal='network'),
    Fact(no_internet=True)
) 

# Run the engine
print("\n--- Test Case 2 Results ---")
engine.run()
```
